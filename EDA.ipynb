```python
# Plot histogram of award years
plt.figure(figsize=(10, 5))
sns.histplot(df['award_year'].dropna(), bins=30, kde=False)
plt.title('Distribution of Award Years')
plt.xlabel('Award Year')
plt.ylabel('Frequency')
plt.tight_layout()
plt.savefig('award_year_distribution.png')
```

```python
# Plot box plot of adjusted prize amounts by category
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='category', y='prize_amount_adjusted')
plt.title('Prize Amount Adjusted by Category')
plt.xlabel('Category')
plt.ylabel('Prize Amount Adjusted')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('prize_amount_adjusted_boxplot.png')
```

```python
numeric_df = df.select_dtypes(include=[np.number])
if numeric_df.shape[1] >= 4:
    plt.figure(figsize=(12,10))
    correlation = numeric_df.corr()
    sns.heatmap(correlation, annot=True, fmt='.2f', cmap='coolwarm')
    plt.title('Correlation Heatmap of Numeric Features')
    plt.tight_layout()
    plt.savefig('correlation_heatmap.png')
```

```python
# Which country produce the most nobel winners over the years?
country_year = (
    df.groupby(["award_year", "birth_country"])
      .size()
      .reset_index(name="winner_count")
)

top_countries = (
    df["birth_country"]
    .value_counts()
    .head(10)
    .index
)

country_year_top10 = country_year[
    country_year["birth_country"].isin(top_countries)
]

fig, ax = plt.subplots(figsize=(14, 7))
for country in top_countries:
    country_data = country_year_top10[
        country_year_top10["birth_country"] == country
    ]
    ax.plot(
        country_data["award_year"],
        country_data["winner_count"],
        label=country
    )
ax.set_xlabel("Award Year")
ax.set_ylabel("Number of Nobel Winners")
ax.set_title("Nobel Winners by Birth Country Over Time")
ax.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.show()
```

```python
# cumulative over time

country_year["cumulative_winners"] = (
    country_year
    .groupby("birth_country")["winner_count"]
    .cumsum()
)

top_countries = (
    df["birth_country"]
    .value_counts()
    .head(10)
    .index
)

country_year_top10 = country_year[
    country_year["birth_country"].isin(top_countries)
]

fig, ax = plt.subplots(figsize=(14, 7))
for country in top_countries:
    country_data = country_year_top10[
        country_year_top10["birth_country"] == country
    ]
    
    ax.plot(
        country_data["award_year"],
        country_data["cumulative_winners"],
        label=country
    )

ax.set_xlabel("Award Year")
ax.set_ylabel("Cumulative Nobel Winners")
ax.set_title("Cumulative Nobel Winners by Birth Country Over Time")
ax.legend(bbox_to_anchor=(1.05, 1), loc="upper left")

plt.show()
```
